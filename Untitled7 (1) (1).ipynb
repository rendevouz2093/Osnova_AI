```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

data = [
    [0.20, 0.10, 0.80, 0.10, 1],
    [0.30, 0.20, 0.70, 0.20, 1],
    [0.40, 0.15, 0.90, 0.15, 1],
    [0.50, 0.30, 0.80, 0.10, 1],
    [0.60, 0.25, 0.75, 0.20, 1],
    [0.70, 0.35, 0.85, 0.10, 1],
    [0.80, 0.40, 0.90, 0.15, 1],
    [0.25, 0.50, 0.70, 0.40, 1],
    [0.35, 0.45, 0.80, 0.35, 1],
    [0.45, 0.55, 0.75, 0.45, 1],
    [0.55, 0.50, 0.85, 0.30, 1],
    [0.65, 0.60, 0.80, 0.40, 1],
    [0.20, 0.80, 0.40, 0.75, 0],
    [0.30, 0.70, 0.35, 0.80, 0],
    [0.40, 0.85, 0.45, 0.70, 0],
    [0.50, 0.75, 0.30, 0.85, 0],
    [0.60, 0.90, 0.40, 0.80, 0],
    [0.70, 0.80, 0.35, 0.90, 0],
    [0.80, 0.85, 0.45, 0.75, 0],
    [0.90, 0.70, 0.30, 0.85, 0],
    [0.25, 0.90, 0.50, 0.80, 0],
    [0.45, 0.95, 0.40, 0.90, 0],
    [0.65, 0.85, 0.30, 0.95, 0],
    [0.85, 0.90, 0.35, 0.85, 0]
]

columns = ['Vibration', 'Slip', 'SurfaceGrip', 'ObstacleRate', 'Stable']
df = pd.DataFrame(data, columns=columns)

train_data = df.iloc[:18]
test_data = df.iloc[18:]

X_train = train_data[['Vibration', 'Slip', 'SurfaceGrip', 'ObstacleRate']]
y_train = train_data['Stable']

X_test = test_data[['Vibration', 'Slip', 'SurfaceGrip', 'ObstacleRate']]
y_test = test_data['Stable']

base_model = LogisticRegression()
base_model.fit(X_train, y_train)
base_preds = base_model.predict(X_test)

print("Базалық модель нәтижелері")
print("Болжам:", base_preds)
print("Нақты:", y_test.values)
print("Дәлдік:", accuracy_score(y_test, base_preds))

nn_model = MLPClassifier(hidden_layer_sizes=(5,), max_iter=1000, random_state=42)
nn_model.fit(X_train, y_train)
nn_preds = nn_model.predict(X_test)

print("Нейрондық желі нәтижелері")
print("Болжам:", nn_preds)
print("Дәлдік:", accuracy_score(y_test, nn_preds))
```
